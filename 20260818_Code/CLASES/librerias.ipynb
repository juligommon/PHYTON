''' Introducción a Librerías en Python

1. ¿Qué es una Librería (o Paquete)? Hasta ahora, hemos trabajado con el “Python puro” (o Built-in).

La Analogía del Vehículo: Imagina que Python puro es un coche que compras de fábrica. Viene con lo básico: motor, llantas, asientos y un funcionamiento perfecto para moverte (hacer cálculos, bucles, funciones).

Pero, ¿qué pasa si quieres ir a la nieve? Necesitas llantas especiales. ¿Y si quieres acampar? Necesitas un remolque.

Una librería es exactamente eso: un paquete de herramientas, funciones y código que alguien más ya programó y que nosotros podemos “instalarle” a nuestro Python para darle superpoderes específicos sin tener que programarlos desde cero.

2. El Gestor de Paquetes: pip install
Antes de usar una herramienta nueva, tenemos que descargarla e instalarla en nuestra computadora o entorno. Para esto usamos PIP (Pip Installs Packages), que es como la “App Store” de Python.

¿Cómo se usa? Esto no se escribe en el script normal de Python, sino en la Consola/Terminal (en VS Code) o usando un símbolo de exclamación ! en una celda de código (en Google Colab).

3. Activando las herramientas: import
Instalar la librería es como comprar la caja de herramientas y dejarla en el garaje. Para usarla en nuestro código hoy, tenemos que abrirla. Eso lo hacemos con la palabra reservada import.

Existen tres formas principales de importar, dependiendo de qué tan grande sea la librería y qué tanto necesitemos de ella.

A. Importación Básica

(import libreria)

Trae absolutamente toda la caja de herramientas. Para usar una herramienta, debemos escribir el nombre de la caja primero, seguido de un punto ..

B. Importación con Alias

(import libreria as alias)

¡El estándar de la industria! A veces los nombres de las librerías son largos. Para no escribirlos muchas veces en nuestro código, les ponemos un apodo corto. En datos, hay apodos universales que TODO el mundo respeta.

C. Importación Específica

(from libreria import herramienta)

Si de una caja de herramientas gigante solo necesitamos un destornillador, no tiene sentido cargar toda la caja en la memoria de la computadora.'''

4. 4. Ejemplos de Usos Posibles (El Ecosistema de Datos)

Para que los estudiantes visualicen el poder de lo que viene, estos son los “superpoderes” que vamos a adquirir al usar librerías:

Manipulación de Datos (Pandas / NumPy): Reemplazan a Excel. Permiten abrir archivos de millones de filas, agrupar, filtrar y limpiar datos en milisegundos.

Visualización (Matplotlib / Seaborn): Permiten convertir tablas aburridas en gráficos de líneas, barras, mapas de color y dashboards interactivos.

Conexión a la Web (Requests): Permite que nuestro código de Python entre a una página web o a una API y descargue datos automáticamente (Web Scraping).

Automatización (Os / Shutil): Permite que Python navegue por las carpetas de tu computadora, cree directorios, mueva archivos y organice tu disco duro sin que tú hagas clic.
