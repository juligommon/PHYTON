Introducción a Pandas

1. ¿Qué es Pandas y por qué lo usamos? Si Python es el motor del coche, Pandas es el sistema de navegación y control de mando para los datos. Es la librería estándar a nivel mundial para la manipulación y análisis de datos estructurados.

La Analogía: Pandas es como un Excel con superpoderes.

En Excel, si intentas abrir un archivo con 2 millones de filas, la computadora se congela.

En Pandas, puedes abrir, limpiar y transformar esos 2 millones de filas en un par de segundos usando código, sin necesidad de usar el mouse ni tener la tabla visible todo el tiempo.

2. Los Dos Conceptos Clave: Series y DataFrames Para trabajar con Pandas, solo necesitamos entender dos estructuras visuales:

La Serie (Series): Es una sola columna de datos. Imagina que extraes solo la columna “Edades” de tu Excel. Eso es una Serie.

El DataFrame (DataFrame): Es una tabla completa con filas y múltiples columnas. Es el equivalente exacto a una hoja de cálculo. En código, solemos llamarlo simplemente df.

3. Instalación e Importación (El Estándar) Como vimos en la teoría de librerías, Pandas no viene preinstalado en Python puro.

Una vez instalada, la comunidad mundial de datos creó un “lenguaje” universal que siempre es respetado por todos en el mundo al crear la tabla.

4. El DataFrame: La estrella (Objeto) Aquí es donde la magia ocurre. Antes de cualquier cosa, hay que tener datos. Pandas tiene herramientas para leer casi cualquier formato (CSV, Excel, SQL) y convertirlo automáticamente en un DataFrame (una tabla).

5. Exploración Básica (Viendo los datos "a ciegas") Como en Python no tenemos la cuadrícula visual de Excel abierta permanentemente, Pandas nos da comandos para dar "vistazos" rápidos a la tabla que acabamos de cargar

```python
import pandas as pd

# 1. A partir de una lista simple
ventas_lista = [150.50, 200.00, 99.99]

# Convertimos la lista en una Serie de Pandas
serie_ventas = pd.Series(ventas_lista)
print("--- Serie Básica ---")
print(serie_ventas)
```

```python
# Diccionario donde cada clave es una columna
datos_clientes = {
    "ID": [101, 102, 103],
    "Nombre": ["Ana", "Luis", "Carlos"],
    "Edad": [28, 34, 25],
    "Estado": ["Activo", "Inactivo", "Activo"]
}

print(datos_clientes)

# Convertimos el diccionario a DataFrame
df_clientes = pd.DataFrame(datos_clientes)

print("--- DataFrame desde Diccionario ---")
print(df_clientes)
```

```python
import pandas as pd

# Una lista donde cada elemento (diccionario) es una fila completa
transacciones = [
    {"Transaccion": "T-01", "Monto": 500, "Sucursal": "Norte"},
    {"Transaccion": "T-02", "Monto": 1200, "Sucursal": "Sur"},
    {"Transaccion": "T-03", "Monto": 300, "Sucursal": "Este"}
]

# Pandas es tan inteligente que lo mapea automáticamente
df_transacciones = pd.DataFrame(transacciones)

print("--- DataFrame desde Lista de Diccionarios ---")
print(df_transacciones)
```
