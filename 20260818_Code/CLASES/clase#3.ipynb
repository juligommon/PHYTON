```python
#Condicionales con Python
Las estructuras condicionales permiten ejecutar bloques de código específicos solo si se cumple una determinada condición booleana 
(True o False).
============ Operadores de comparación ============

== igual a

!= Diferente de

< Menor que

> mayor que

<= menor o igual a

>= mayor o igual a

============ Operadores lógicos ============

Permiten combinar múltiples condiciones en una sola expresión.

and: Devuelve True solo si ambas condiciones son verdaderas.

or: Devuelve True si al menos una de las condiciones es verdadera.

not: Invierte el valor booleano (convierte True en False y viceversa).

===== Solo If: La regla estricta =====

Lo usamos cuando solo nos interesa ejecutar una acción si la condición es verdadera.
Si es falsa, el programa simplemente ignora el bloque y continúa su camino sin interrumpir el proceso.
```

```python
pais = 'Colombia'
tu_pais = input('En que pais naciste')

if pais == tu_pais:
    print("La tierra del sagrado corazon")

# print(f"Me alegro que hayas nacido en {tu_pais}")
```

```python
# if - else: La bifurcación (Plan A o Plan B) ===

Aquí le damos a Python dos caminos mutuamente excluyentes.
Si la condición principal se cumple, ejecuta el Plan A. Si no se cumple, cae automáticamente en el Plan B (el else).
```

```python
nota_aprobacion = 4.0
nota_estudiante = float(input('Digite la nota del estudiante: '))

if nota_estudiante >= nota_aprobacion:
    print("Aprobaste")
else:
    print("Reprobaste")
```

```python
# if - elif - else: El clasificador =====
Nos permite evaluar múltiples escenarios en orden.
 Python revisa el if; si es falso, salta al primer elif, y así sucesivamente. 
 El else final actúa como una red de seguridad para capturar todo lo que no encajó en las reglas anteriores.
```

```python
NPS = int(input('En una escala de 1 a 10 que tanto recomiendas el producto o servicio?'))

if NPS >= 9:
    print('Promotor')
elif NPS > 6:
    print('Neutro')
elif NPS >= 6 and NPS > 0:
    print('Detractor')
else:
    print('Revisar urgente')
```

```python
# Múltiples condiciones (and / or): El filtro avanzado ===

En el análisis diario, rara vez evaluamos una sola variable. Usamos operadores lógicos para combinar varias 
reglas de negocio en una sola validación.
and (Y lógico): Obliga a que todas las condiciones sean verdaderas para ejecutar el código.
or (O lógico): Permite ejecutar el código si al menos una condición es verdadera.
```

```python
edad = 28
salario = 3500
reportes_datacredito = 'no'

tu_edad = int(input('Digite su edad: '))
ti_salario = float(input('Digite su salario: '))
tu_reportes_datacredito = input('Tiene reportes en Datacredito? (si/no): ')

# Evaluamos tres variables al mismo tiempo
if edad >= tu_edad and salario > tu_salario and reportes_datacredito == 'tu_reportes_datacredito':
    print("Cliente aprobado para producto financiero.")
else:
    print("El cliente no cumple con los requisitos.")
```
