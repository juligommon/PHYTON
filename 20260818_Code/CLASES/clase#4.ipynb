```python
# El ciclo continúa hasta que el usuario decida salir
activo = True

while activo:
    opcion = input("Escribe 'salir' para terminar o cualquier tecla para continuar: ").lower()
    
    if opcion == "salir":
        activo = False  # Cambia el valor de la variable inicial y detiene el bucle en la próxima evaluación
    else:
        print('Procesando datos...{opcion}')
        print('terminamos')

# print("Terminamos")
```

## El Bucle for: Automatización Masiva

### A diferencia del while, el for sabe exactamente cuántas veces se va a repetir. Sirve para recorrer colecciones (como las listas y diccionarios que ya vimos) elemento por elemento, haciendo que procesos largos se resuelvan en tres líneas de código.

### Iterando colecciones: Perfecto para ejecutar exactamente la misma transformación en una lista de múltiples archivos.

### Usando range(): Genera secuencias numéricas, ideal para repetir una extracción un número específico de veces.

```python
# sintaxis
#for i in range(0, 100, 10):
#print(i)

for alias in range(1,10,2):
    print(alias)
```

```python
# Imprime los múltiplos de 3 entre 3 y 15
for numero in range(3, 16, 3):
    print(f"Múltiplo de 3: {numero}")
```

```python
# for para recorrer listas.
tablas = ["hechos_ventas", "dim_clientes", "dim_tiempo", "dim_pais"]

for tabla in tablas:
    print(f"Ejecutando limpieza en la tabla: {tabla}")
```

### Para llegar a un nivel intermedio, necesitamos controlar el comportamiento interno del ciclo frente a imprevistos.

### break (Freno de emergencia): Detiene el ciclo por completo. Útil si detectamos un error crítico de conexión y debemos abortar todo el proceso.

### continue (Saltar registro): Detiene solo la iteración actual y pasa inmediatamente a la siguiente. Excelente para ignorar registros nulos o dañados sin detener el resto del procesamiento.

```python
# Buscar un producto específico en el inventario
inventario = ["camisa", "pantalón", "zapatos", "chaqueta", "gorra"]
producto_buscado = "zapatos"

for producto in inventario:
    print(f"Revisando: {producto}")
    if producto == producto_buscado:
        print(f"¡Éxito! Se encontró el artículo: {producto}.")
        break  # Detiene el bucle aquí (ya no revisa 'chaqueta' ni 'gorra')
```

```python
# Monitoreo de temperatura: detener el sistema si supera los 40 grados
lecturas_temperatura = [22, 25, 28, 40, 30, 24]

for temperatura in lecturas_temperatura:
    if temperatura > 40:
        print(f"¡ALERTA! Temperatura crítica detectada ({temperatura}°C). Apagando sistema...")
        break  # Interrumpe el bucle de inmediato por seguridad
    print(f"Temperatura normal: {temperatura}°C")
```

```python
# Enviar notificaciones solo a correos válidos
correos = ["ana@email.com", "invalido.com", "carlos@email.com", "", "sofia@email.com"]

for email in correos:
    # Si falta el "@" o está vacío, saltamos este registro
    if "@" not in email or email == "":
        print("Registro omitido: formato incorrecto.")
        continue  # Salta directo al siguiente correo sin ejecutar el print de abajo

    print(f"Enviando notificación a: {email}")
```
