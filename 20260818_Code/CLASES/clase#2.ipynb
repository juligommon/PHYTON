```python
#Listas: Colecciones ordenadas y modificables. Son tu herramienta del día a día para almacenar series, 
#como rutas de archivos a procesar. Ejemplo: sucursales = [“Bogotá”, “Medellín”]

lista_paises = [
    "Argentina",
    "Bolivia",
    "Ucrania",
    "España",
    "Lituania",
    "Eslovenia",
    "Grecia",
    "Colombia",
    "Chile"
]

print(lista_paises[-2])
```

```python
lista_paises.append("Perú")
```

```python
lista_paises.reverse()
print(lista_paises)
```

```python
print(lista_paises)
```

```python
#append(): Añade un elemento al final.
#insert(): Pone un elemento en un lugar fijo.
#remove(): Borra un elemento específico.
#pop(): Saca un elemento de una posición.
#sort(): Ordena los elementos.
#reverse(): Invierte el orden de los elementos.
```

```python
#Diccionarios (dict): Almacenan información en pares de clave: valor. Tienen la misma estructura que un archivo JSON.
#  Ejemplo: producto = {“id”: 101, “categoria”: “Operaciones”}

paises_capitales = {
    'Colombia': 'Bogotá',
    'Bolivia': 'La Paz',
    'Chile': 'Santiago de Chile',
    'Argentina': 'Buenos Aires',
    'Perú': 'Lima'
}
```

```python
print(paises_capitales)
```

```python
paises_capitales.update({'Ecuador': 'Quito'})
```

```python
print(paises_capitales)
```
