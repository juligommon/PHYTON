Matplotlib para Visualización de Datos

Fundamentos y Dominio de Matplotlib para Visualización de Datos

¿Qué es Matplotlib? Es la librería estándar y más potente en Python para la creación de gráficos estáticos, animados e interactivos. Su arquitectura está diseñada para replicar el comportamiento de visualización de MATLAB, permitiendo un control absoluto sobre cada elemento visual de una figura.

La Jerarquía de Objetos (Figure y Axes): A diferencia de un simple dibujo, Matplotlib separa la ventana o lienzo general (Figure) de los espacios de trazado individuales (Axes). Comprender esta estructura es vital para construir gráficos profesionales y personalizados.

Aplicaciones y Usos en el Mercado:

Ciencia de Datos y Exploración (EDA): Identificación visual de tendencias, correlaciones y valores atípicos (outliers) en datasets masivos.

Reportes Ejecutivos y Dashboards: Generación de gráficos claros (líneas, barras, dispersión) para la toma de decisiones gerenciales.

Monitoreo y Modelos de Machine Learning: Visualización de métricas de rendimiento como curvas de aprendizaje, pérdida (loss curves) y matrices de confusión.

Temas Principales para Dominar la Librería

La Anatómia de un Gráfico: Uso de etiquetas (xlabel, ylabel), títulos (title), leyendas (legend) y rejillas (grid) para garantizar legibilidad profesional.

Gráficos de Líneas y Series Temporales: Ideales para mostrar evolución cronológica de variables continuas.

Gráficos de Barras y Columnas: Perfectos para comparar categorías discretas o volúmenes agrupados.

Diagramas de Dispersión (Scatter Plots) y Histogramas: Esenciales para analizar correlaciones estadísticas y distribuciones de frecuencias.

Subplots (Múltiples Paneles): Técnicas para organizar varios gráficos dentro de una misma ventana de visualización usando plt.subplots().

1. Gráficos de Líneas (Evolución Temporal)

Explicación: Se utilizan principalmente para representar cambios y tendencias de una variable numérica a lo largo del tiempo o secuencias continuas.

```python
%pip install matplotlib
import matplotlib.pyplot as plt
```

```python
meses = ['Ene', 'Feb', 'Mar', 'Abr', 'May']
ventas = [12000, 15000, 11000, 18000, 22000]
plt.figure(figsize=(5 , 3))
plt.plot(meses,ventas,
        marker='o',
        color="#041F50",
        linewidth = 1,
        linestyle=':')
plt.title('Evolución de Ventas Semestrales')
plt.xlabel('Meses', fontsize=8 )
plt.ylabel('Ventas', fontsize=12)
plt.grid(visible=True, 
    color='gray', 
    linestyle=':', 
    linewidth=0.3,  
    alpha=0.9)
plt.show()
```

2. Gráficos de Barras (Comparación Categórica)

Explicación: Permiten contrastar magnitudes o volúmenes entre diferentes categorías discretas de forma clara y directa.

```python
categorias = ['Accesorios', 'Periféricos', 'Electrónica', 'Monitores']
unidades = [150, 80, 200, 95]

plt.figure(figsize=(5, 3))
plt.bar(categorias, unidades, color=['orange', 'green', 'purple', 'blue'])
plt.title('Unidades Vendidas por Categoría' , fontsize = 12 , color="#F669E8")
plt.xlabel('Categoría')
plt.ylabel('Cantidad Total')
plt.show()
```

3. Diagramas de Dispersión (Scatter Plot)

Explicación: Relacionan dos variables cuantitativas para identificar correlaciones, agrupamientos o posibles valores atípicos (outliers).

```python
precios = [50000, 120000, 300000, 450000, 1200000]
cantidades = [10, 8, 5, 4, 2]

plt.style.use('seaborn-v0_8-dark-palette') 
plt.figure(figsize=(8, 3))
plt.scatter(precios, cantidades, color='orange', s=24 )
plt.title('Relación entre Precio y Cantidad Vendida')
plt.xlabel('Precio Unitario')
plt.ylabel('Cantidad')
plt.grid(True)
plt.show()
```

4. Histogramas (Distribución de Frecuencias)

Explicación: Agrupan datos numéricos continuos en intervalos (bins) para mostrar cómo se distribuyen las frecuencias de los valores.

```python
import matplotlib.pyplot as plt
import numpy as np

# Generación de datos aleatorios simulando precios
precios_dataset = np.random.normal(90000, 20000, 2000)

plt.figure(figsize=(10,4))
plt.hist(precios_dataset, bins=50, color='skyblue' , rwidth=0.5 )
plt.title('Distribución de Frecuencia de Precios')
plt.xlabel('Rango de Precios')
plt.ylabel('Frecuencia')
plt.show()
```

5. Subplots (Múltiples Paneles en una sola Figura)

Explicación: Permiten estructurar una cuadrícula de gráficos independientes utilizando plt.subplots(), ideal para comparar múltiples métricas en un mismo informe visual.

```python
import matplotlib.pyplot as plt
plt.style.use('ggplot')
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 4))

# Gráfico 1 (Izquierda)
ax1.plot([1, 2, 3], [10, 20, 25], color='blue')
ax1.set_title('Tendencia A')

# Gráfico 2 (Derecha)
ax2.bar(['A', 'B', 'C'], [5, 12, 8], color='orange')
ax2.set_title('Comparativa B')

plt.tight_layout()
plt.show()
```
