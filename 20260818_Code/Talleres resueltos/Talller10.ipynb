Ejercicio 24: Agregaciones Múltiples con .agg()

Contexto: El departamento financiero requiere un reporte avanzado que detalle simultáneamente el ingreso total y el precio promedio por categoría.

Ejercicio 28: Uso de Bloque try-except para Conversión de Tipos

Contexto: Al transformar columnas de texto numérico a formato flotante o entero, pueden surgir errores por caracteres inesperados.

Ejercicio 30: Consolidación y Resumen Final con Funciones y Bucles

Contexto: Crearemos un bloque de script que combine lógica de funciones personalizadas y bucles para resumir estadísticas clave del dataset limpio.

Instrucción: Define una función que reciba un DataFrame y devuelva el ingreso total acumulado ("venta_total"). Utiliza un bucle for para iterar sobre las ciudades únicas del DataFrame, imprimiendo el ingreso total generado en cada ciudad específica calculado mediante un filtro dinámico.

```python
# Calcular la venta total
df["venta_total"] = df["precio"] * df["cantidad"]


# Crear una función para calcular el ingreso total
def calcular_ingreso_total(datos):
    return datos["venta_total"].sum()


# Obtener las ciudades únicas
ciudades = df["ciudad"].unique()


# Recorrer cada ciudad
for ciudad in ciudades:

    # Filtrar los datos de la ciudad actual
    df_ciudad = df[df["ciudad"] == ciudad]

    # Calcular el ingreso total de esa ciudad
    ingreso = calcular_ingreso_total(df_ciudad)

    # Mostrar el resultado
    print("Ciudad:", ciudad, "| Ingreso total:", ingreso)
```

```python
def clasificar_notas(notas):
    aprobados = 0
    no_aprobados = 0

    for nota in notas:
        if nota >= 3.0:
            aprobados += 1
        else:
            no_aprobados += 1

    return (aprobados, no_aprobados)

print(clasificar_notas([4.5, 2.8, 3.0, 1.9, 3.7]))
```
