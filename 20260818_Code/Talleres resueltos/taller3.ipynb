```python
#1. El Monitor de Calidad
#Contexto: Eres responsable de revisar la calidad de una extracción de datos.
#Instrucción: Escribe un programa que solicite al usuario ingresar la cantidad de “valores nulos” encontrados en una tabla. 
#Utiliza un condicional simple (if) para evaluar si la cantidad es mayor a 0. Si es así, imprime el mensaje:
#“Alerta: Se detectaron registros incompletos. Revisar origen.”

valores_nulos = int(input("Ingrese la cantidad de valores nulos: "))

if valores_nulos > 0:
    print("Alerta: Se detectaron registros incompletos. Revisar origen.")
```

```python
#2. El Enrutador de Archivos
#Contexto: Tienes un script que solo sabe procesar archivos en formato CSV.
#Instrucción: Pide al usuario que escriba la extensión de un archivo (por ejemplo: “csv”, “excel”, “json”).
#Si la extensión ingresada es exactamente igual (==) a “csv”, imprime:
#“Formato correcto. Iniciando lectura…”
#Si ingresa cualquier otra cosa, el programa debe caer en el else e imprimir:
#“Error: Formato no soportado, por favor convierta el archivo.”

extension = input("Ingrese la extensión del archivo: ")

if extension == "csv":
    print("Formato correcto. Iniciando lectura...")
else:
    print("Error: Formato no soportado, por favor convierta el archivo.")
```

```python

#El Clasificador de Satisfacción - NPS
#Contexto: El equipo de marketing necesita categorizar las respuestas de una encuesta de satisfacción (del 0 al 10).
#Instrucción: Solicita al usuario una calificación entera.
# Si la nota es mayor o igual a 9, imprime “Cliente Promotor”.
#Si la nota es mayor o igual a 7 (pero no llegó a 9), imprime “Cliente Neutro”.
# Para cualquier nota por debajo de 7, imprime “Cliente Detractor”.

nota = int(input("Ingrese una calificación: "))

if nota >= 9:
    print("Cliente Promotor")
elif nota >= 7:
    print("Cliente Neutro")
else:
    print("Cliente Detractor")
```

```python
#4. Validación de Asistencia a Eventos
#Contexto: Estás construyendo un flujo automatizado para capturar registros de un formulario de eventos y mandarlos a una base de datos.
#Instrucción: Pide al usuario que ingrese su edad y su estado de registro (escribiendo “confirmado” o “pendiente”).
#  Utiliza el operador and para validar que el asistente tenga 18 años o más y que su estado sea “confirmado”.
#Si cumple ambas, imprime:
#“Registro válido. Procesando en base de datos”
#De lo contrario, imprime:
#“Registro retenido”.

edad = int(input("Ingrese su edad: "))
estado = input("Ingrese el estado de registro (confirmado o pendiente): ")

if edad >= 18 and estado == "confirmado":
    print("Registro válido. Procesando en base de datos")
else:
    print("Registro retenido")
```

```python
#5. Filtro de Sucursales
#Contexto: Estás filtrando ventas y necesitas separar las de las ciudades principales.
#Instrucción: Solicita al usuario ingresar el nombre de una ciudad. Utiliza el operador or para evaluar si la ciudad es
#  “Bogotá” o “Medellín”.
#Si alguna de las dos es cierta, imprime:
#“Venta clasificada como: Región Principal”.
#Si es cualquier otra ciudad, imprime:
#“Venta clasificada como: Región Secundaria”.

ciudad = input("Ingrese el nombre de la ciudad: ")

if ciudad == "Bogotá" or ciudad == "Medellín":
    print("Venta clasificada como: Región Principal")
else:
    print("Venta clasificada como: Región Secundaria")
```

```python
#6. El Liquidador de Comisiones (Lógica Combinada)
#Contexto: Eres el analista encargado de calcular si un vendedor recibe comisión este mes, combinando varias reglas de negocio.
#Instrucción: Solicita las ventas totales del mes (número) y la antigüedad del vendedor en años (número).
#* Si las ventas superan los 10000 y la antigüedad es mayor a 2 años, gana comisión del 10%.
#* Si las ventas superan los 10000 pero tiene 2 años o menos, o si las ventas están entre 5000 y 10000 sin importar la antigüedad, gana comisión del 5% (usa elif).
#* Si no cumple nada de lo anterior, no hay comisión. Muestra el mensaje correspondiente.

ventas = float(input("Ingrese las ventas totales del mes: "))
antiguedad = float(input("Ingrese la antigüedad del vendedor en años: "))

if ventas > 10000 and antiguedad > 2:
    print("Gana comisión del 10%")
elif ventas > 10000 and antiguedad <= 2 or ventas >= 5000 and ventas <= 10000:
    print("Gana comisión del 5%")
else:
    print("No hay comisión")
```
