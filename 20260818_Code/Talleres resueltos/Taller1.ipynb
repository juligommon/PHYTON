```python
# Calculadora de Edad (Calentamiento). 
# Escribe un programa que pregunte al usuario su año de nacimiento y el año actual. 
# El programa debe calcular su edad y mostrar el siguiente mensaje en pantalla: "Tienes [edad] años o los cumplirás este año".

año_actual = int(input("¿En qué año estamos? "))
año_nacimiento = int(input("¿En qué año naciste? "))

edad = año_actual - año_nacimiento

print("Tienes", edad, "años o los cumplirás este año")
```

```python
# 2 Promedio de Satisfacción al Cliente (CX)
#Imagina que estás procesando los resultados de un formulario de encuestas de servicio al cliente.
# El programa debe pedir al usuario que ingrese las calificaciones de 3 encuestas diferentes (números enteros del 1 al 5).
#Calcula el promedio exacto de estas tres calificaciones (con decimales) y muéstralo en pantalla.

calificacion1 = int(input("Ingresa la primera calificación (1 al 5): "))
calificacion2 = int(input("Ingresa la segunda calificación (1 al 5): "))
calificacion3 = int(input("Ingresa la tercera calificación (1 al 5): "))

promedio = (calificacion1 + calificacion2 + calificacion3) / 3

print("El promedio de satisfacción es:", promedio)
```

```python
# 3	Auditoría de Tiempos de Ejecución 
# Tienes un script automatizado que se ejecuta en el servidor y tarda cierta cantidad de minutos en procesar la data.
#  Construye un programa que pida al usuario ingresar el total de minutos que tardó el proceso. 
# Utilizando los operadores de división entera (//) y módulo (%), convierte ese tiempo a horas y minutos.
#  Ejemplo: Si el usuario ingresa 130, el programa debe imprimir: "El proceso tardó 2 horas y 10 minutos"

minutos = int(input("Ingresa el total de minutos que tardó el proceso: "))

horas = minutos // 60
minutos_restantes = minutos % 60

print("El proceso tardó", horas, "horas y", minutos_restantes, "minutos")
```

```python
# 4 Generador de Correos Corporativos Estás ayudando a Recursos Humanos a limpiar la base de  datos de los nuevos empleados. Crea un programa que solicite al usuario:
# Su primer nombre.
# Su primer apellido.
# El nombre de la empresa. 
# El programa debe concatenar (unir) estos textos para imprimir el correo electrónico oficial. 
# Ejemplo de salida: "Tu correo asignado es: nombre.apellido@empresa.com"

nombre = input("Ingresa tu primer nombre: ")
apellido = input("Ingresa tu primer apellido: ")
empresa = input("Ingresa el nombre de la empresa: ")

correo = nombre + "." + apellido + "@" + empresa + ".com"

print("Tu correo asignado es:", correo)
```
