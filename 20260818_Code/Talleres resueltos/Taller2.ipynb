```python
#El Laboratorio de Transformación (Manipulación de Listas)
#Eres el encargado del inventario en un e-commerce. Tienes la siguiente lista inicial de productos más vendidos hoy:
#inventario = [“Laptop”, “Monitor”, “Teclado”, “Mouse”]
#Escribe el código paso a paso para realizar las siguientes transformaciones, imprimiendo la lista final al terminar todo el proceso:
#1. Añade unos “Audífonos” al final del inventario.
#2. Inserta una “Tablet” exactamente en la segunda posición de la lista (índice 1).
#3. Borra el “Teclado” porque se agotó el stock usando el método que busca por nombre.
#4. Saca el último elemento de la lista (para enviarlo a bodega) usando el método correspondiente y guárdalo en una variable llamada
#  producto_enviado.
#5. Ordena el inventario restante alfabéticamente (de la A a la Z).
#6. Dale vuelta a la lista para verla en orden Z a A.
# 
# 
# Lista inicial
inventario = ["Laptop", "Monitor", "Teclado", "Mouse"]

# 1. Añadir "Audífonos" al final
inventario.append("Audífonos")

# 2. Insertar "Tablet" en la segunda posición (índice 1)
inventario.insert(1, "Tablet")

# 3. Borrar "Teclado" buscando por su nombre
inventario.remove("Teclado")

# 4. Sacar el último elemento y guardarlo en producto_enviado
producto_enviado = inventario.pop()

# 5. Ordenar el inventario restante de la A a la Z
inventario.sort()

# 6. Darle vuelta para verlo de la Z a la A
inventario.reverse()

# Imprimir la lista final
print(inventario)
```
