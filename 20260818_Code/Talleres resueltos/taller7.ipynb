```python
''' Ejercicio 1: Usando herramientas nativas (Librería math)
Contexto: Necesitas calcular la raíz cuadrada del área de un terreno para un análisis geoespacial.
Instrucción: Importa la librería math y utiliza su función sqrt() para calcular la raíz cuadrada de 144.
 Imprime el resultado con un mensaje claro.'''

import math

raiz = math.sqrt(144)

print("La raíz cuadrada de 144 es:", raiz)
```

```python
'''Ejercicio 2: Creando una Serie en Pandas
Contexto: Tienes un listado de precios de productos y quieres organizarlos con sus respectivas etiquetas de categoría.
Instrucción: Importa pandas como pd. Crea una lista con los precios [1200, 450, 89] y una lista de etiquetas 
["Laptop", "Mouse", "Teclado"] para usarlas como índice. Convierte esto en una pd.Series e imprímela.'''

import pandas as pd

precios = [1200, 450, 80]
etiquetas = ["Laptop", "Mouse", "Teclado"]

serie = pd.Series(precios, index=etiquetas)

print(serie)
```

```python
'''Ejercicio 3: Creando un DataFrame desde un Diccionario
Contexto: Necesitas armar una tabla rápida con información de sucursales para una presentación ejecutiva.
Instrucción: Importa pandas como pd. Crea un diccionario con dos columnas: "Ciudad" ["Bogotá", "Medellín", "Cali"] 
y "Ventas" [5000, 4200, 3100]. Convierte el diccionario en un DataFrame y muéstralo en consola.'''

import pandas as pd

datos = {
    "Ciudad": ["Bogotá", "Medellín", "Cali"],
    "Ventas": [35000, 42000, 31000]
}

df = pd.DataFrame(datos)

print(df)
```

```python
'''Ejercicio 4: DataFrame desde una Lista de Diccionarios (Formato API)
Contexto: Recibes datos de transacciones en formato de diccionarios independientes y debes estructurarlos en una tabla limpia.
Instrucción: Importa pandas as pd. Crea una lista con tres diccionarios, cada uno con las claves "ID" y "Estado". 
Conviértela en un DataFrame e imprímela.'''

import pandas as pd

transacciones = [
    {"ID": 1, "Estado": "Completada"},
    {"ID": 2, "Estado": "Pendiente"},
    {"ID": 3, "Estado": "Cancelada"}
]

df = pd.DataFrame(transacciones)

print(df)
```
