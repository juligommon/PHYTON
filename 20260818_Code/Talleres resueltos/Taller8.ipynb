Ejercicio 4: Filtrado Condicional de Registros

Contexto: La gerencia te solicita aislar exclusivamente aquellas transacciones comerciales en donde la cantidad de unidades vendidas sea mayor o igual a 8.

Instrucción: Aplica un filtro condicional sobre el DataFrame principal para conservar solo las filas donde la columna "cantidad" sea >= 8. Guarda este resultado en df_volumen y muestra cuántas transacciones cumplieron con esta condición usando la propiedad .shape.

Ejercicio 5: Creación de una Columna Calculada (Venta Total)

Contexto: El dataset cuenta con el precio unitario y las unidades vendidas, pero carece de la columna que arroje el valor monetario total de cada transacción.

Instrucción: Crea una nueva columna en tu DataFrame llamada “venta_total” multiplicando la columna “precio” por la columna “cantidad”. Luego, muestra las primeras 5 filas de las columnas “id_venta”, “precio”, “cantidad” y “venta_total” para validar el cálculo.

```python
df['venta_total'] = df['precio'] * df['cantidad']
```

```python
df[['id_venta', 'precio', 'cantidad', 'venta_total']].head(5)
```

Ejercicio 6: Limpieza de Valores Nulos (Imputación por Promedio)

Contexto: Como viste en la inspección del Ejercicio 2, la columna “precio” contiene 15 valores nulos (NaN) que podrían corromper nuestros análisis financieros si no se tratan adecuadamente.

Instrucción: Calcula el promedio de la columna “precio” utilizando .mean(). Posteriormente, utiliza ese valor promedio para rellenar los espacios vacíos (NaN) de esa misma columna usando el método .fillna(). Comprueba que el número de nulos en “precio” sea 0 después de la operación.

```python
promedio_precio = df['precio'].mean()
```

```python
print(promedio_precio)
```

```python
df['precio'] = df['precio'].fillna(promedio_precio)
```

```python
df['precio'].isna().sum()
```

Ejercicio 7: Filtrado Combinado con Múltiples Condiciones Lógicas

Contexto: El departamento de análisis de mercado necesita perfilar un grupo específico de transacciones de alto valor y volumen en zonas clave.

Instrucción: Filtra el DataFrame para obtener únicamente aquellas transacciones donde la “cantidad” sea mayor a 5 y la categoría del producto sea exactamente “Accesorios” (puedes usar el operador lógico & y paréntesis para separar las condiciones). Muestra el número total de registros coincidentes.

```python
df['cantidad'] > 5
```

```python
df['categoria'] == 'accesorios'
```

```python
df_accesorios = df[(df["cantidad"] > 5) & (df["categoria"] == "Accesorios")]
df_accesorios.shape
```
