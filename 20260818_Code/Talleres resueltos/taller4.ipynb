```python
# 1. Validador de Asistencia

# Solicita al usuario ingresar el estado de registro de un asistente
# El ciclo debe repetirse mientras el texto ingresado no sea válido.

estado = input("Ingrese el estado del asistente (presente/ausente): ")

while estado != "presente" and estado != "ausente":
    print("Estado no válido. Intente nuevamente.")
    estado = input("Ingrese el estado del asistente (presente/ausente): ")

print("Estado registrado:", estado)
```

```python
#Límite de Reintentos de Conexión
#Simula un inicio de sesión a una base de datos. Pide una contraseña y otorga un máximo de 3 intentos. 
# El ciclo continúa mientras la clave sea incorrecta y la cantidad de intentos sea menor a 3.

clave_correcta = "1234"
intentos = 0

clave = input("Ingrese la contraseña: ")
intentos += 1

while clave != clave_correcta and intentos < 3:
    print("Contraseña incorrecta.")
    clave = input("Ingrese la contraseña nuevamente: ")
    intentos += 1

if clave == clave_correcta:
    print("Acceso concedido.")
else:
    print("Acceso bloqueado. Ha superado el límite de intentos.")
```

```python
# 4. Segmentación de Público

# Dada una lista de edades recolectadas en un formulario,
# recorre cada valor e imprime "Perfil Objetivo"
# solo si la edad es mayor o igual a 18 y menor a 65.

edades = [16, 25, 70, 34, 17, 66, 82]

for edad in edades:
    if edad >= 18 and edad < 65:
        print("Perfil Objetivo")
```

```python
# 5. Limpieza de Errores con continue

# Tiene una lista de transacciones donde algunos valores faltaron y se guardaron como 0 o -1.
# Recorre la lista; si el valor es 0 o -1, sáltalo. Si es correcto, imprímelo.

transacciones = [150, 0, 320, -1, 400, 250, 482]

for transaccion in transacciones:
    if transaccion == 0 or transaccion == -1:
        continue
    print(transaccion)
```

```python
# 6. Freno de Emergencia con break

# Analiza una lista de tiempos de respuesta de un servidor web.
# Si encuentras un pico de latencia estrictamente mayor a 100 milisegundos,
# detén todo el ciclo y lanza una alerta.

tiempos_ms = [15, 20, 105, 30, 22]

for tiempo in tiempos_ms:
    if tiempo > 100:
        print(f"Freno de emergencia activado a {tiempo} milisegundos")
        break
    print(f"Tiempo de respuesta: {tiempo} milisegundos")
```
