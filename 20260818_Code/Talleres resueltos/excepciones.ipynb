'''Manejo de excepciones en Python
Teoría: ¿Por qué atrapar errores? En el mundo real, los datos vienen sucios. Un archivo fuente puede no existir, una columna de precios puede traer el texto “N/A” en lugar de un número, o una métrica puede intentar dividir por cero. El manejo de excepciones nos permite anticipar estos fallos, capturarlos silenciosamente, registrar el incidente y permitir que el programa continúe procesando el resto de los registros.

La Estructura: El Bloque de Seguridad
Para proteger nuestro código, utilizamos un bloque de control específico:

try: Contiene el código principal que queremos ejecutar, sabiendo que existe un riesgo de que falle.

except: Es el “plan de rescate”. Solo se activa si el bloque try colapsa. Podemos tener varios para atrapar errores específicos (ej. ValueError, ZeroDivisionError).

else: Se ejecuta única y exclusivamente si el try tuvo éxito. Es ideal para guardar resultados confirmados.

finally: Se ejecuta siempre, haya ocurrido un error o no. Es vital para tareas de limpieza, como cerrar un archivo abierto o desconectarse del servidor.

Ejemplos de Uso Analítico
Aquí tenemos desde un rescate básico hasta un flujo completo de extracción.
Ejemplo Simple: El Error de Tipo (ValueError)
Ocurre mucho al intentar transformar texto a número.

Lista de excepciones

ZeroDivisionError: Se lanza al intentar dividir un número por cero.

TypeError: Se lanza al aplicar una operación a un objeto de un tipo inapropiado.

ValueError: Se lanza cuando una operación recibe un argumento con el tipo correcto pero un valor inapropiado (por ejemplo, int("abc")).

IndexError: Se lanza al intentar acceder a un índice fuera de rango en una secuencia como una lista.

KeyError: Se lanza cuando una clave no existe en un diccionario.

AttributeError: Se lanza cuando falla la referencia o asignación de un atributo que no existe.

FileNotFoundError: Se lanza al intentar abrir un archivo que no existe en la ruta especificada.'''


```python
# ZeroDivisionError

try:
    calculo = 100 / 0

except ZeroDivisionError as e:
    print(f"Error de cálculo: {e}")
```

```python
# TypeError

try:
    resultado = "Edad: " + 25

except TypeError as e:
    print(f"Error capturado: {e}")
```

```python
# ValueError

try:
    numero = int("pato")

except ValueError as e:
    print(f"Error capturado: {e}")
```

```python
# IndexError

try:
    lista = [10, 20]
    print(lista[5])

except IndexError as e:
    print(f"Error capturado: {e}")
```

```python
# KeyError

try:
    usuario = {"nombre": "Ana", "edad": 30}
    print(usuario["pais"])

except KeyError as e:
    print(f"Error capturado: {e}")
```

```python
# AttributeError

try:
    texto = "Hola Mundo"
    texto.append("!")

except AttributeError as e:
    print(f"Propiedad inválida: {e}")
```

```python
# FileNotFoundError

try:
    with open("archivo_fantasma.txt", "r") as f:
        contenido = f.read()

except FileNotFoundError as e:
    print(f"Archivo no encontrado: {e}")
```

```python
# dos except

try:
    # Intenta convertir una letra a entero
    numero = int(input("numero"))
    numero2 = int(input("numero"))
    print(numero / numero2)

except ValueError:
    print("Error: No se pudo convertir el texto a número.")

except ZeroDivisionError:
    print("Error: División por cero.")
```

```python
# Full ejemplo

try:
    archivo = open("datos.txt", "r")
    contenido = archivo.read()

except FileNotFoundError:
    print("El archivo no existe.")

else:
    print("Lectura exitosa.")
    archivo.close()

finally:
    print("Proceso terminado.")
```
