Ejercicio 1: Carga y Manipulación de Datos

Contexto: Antes de aplicar cualquier transformación o análisis, debemos cargar de manera correcta nuestro dataset garantizando que el delimitador y la codificación operen sin errores, y generar métricas calculadas básicas.

Instrucción: Importa pandas as pd. Lee el archivo ventas_bogota_limpiar.csv usando el separador correcto (sep=”;”) y la codificación “utf-8”. Crea una nueva columna llamada “venta_total” multiplicando la columna “precio” por la columna “cantidad”. Imprime las primeras 3 filas de las columnas “id_venta”, “producto” y “venta_total”.

```python
import pandas as pd
```

```python
df = pd.read_csv('ventas_bogota_limpiar9.csv', sep=';', encoding='utf-8')
```

```python
df['venta_total'] = df['precio'] * df['cantidad']
```

```python
df[["id_venta", "producto", "venta_total"]].head(3)
```

Ejercicio 2: Manipulación de Cadenas (.str)

Contexto: Los datos de texto suelen presentar irregularidades tipográficas o diferencias de mayúsculas y minúsculas que afectan el análisis posterior.

Instrucción: Utiliza el descriptor .str para convertir los valores de la columna “ciudad” a minúsculas y eliminar posibles espacios sobrantes, guardando el resultado en una nueva columna llamada “ciudad_limpia”. Asimismo, estandariza la columna “categoria” reemplazando la palabra “Periféricos” por “Perifericos”. Muestra las primeras 3 filas de estas columnas.

```python
df["ciudad_limpia"] = df["ciudad"].str.lower().str.strip()

df["categoria"] = df["categoria"].str.replace("Periféricos", "Perifericos")

df[["ciudad_limpia", "categoria"]].head(3)
```

Ejercicio 3: Ordenamiento de Datos (sort_values)

Contexto: Para identificar rápidamente las transacciones con mayor impacto financiero dentro del dataset, requerimos estructurar los registros de manera descendente.

Instrucción: Ordena el DataFrame principal en función de la columna “venta_total” de mayor a menor (ascending=False). Muestra las 3 transacciones principales indicando su ID, producto y valor total de venta.

```python
df_ordenado = df.sort_values(by="venta_total", ascending=False)

df_ordenado[["id_venta", "producto", "venta_total"]].head(3)
```

Ejercicio 4: Agrupamiento y Agregaciones (groupby)

Contexto: El negocio requiere un reporte consolidado que resuma el rendimiento económico por categorías de productos.

Instrucción: Agrupa el DataFrame por la columna de tu preferencia y usa el método sum sobre la columna ‘Ventas_total’ para obtener solo el valor de las agrupaciones.

En otra agrupación aplica la misma fórmula pero devuelve la agrupación por promedio de la columna ‘Ventas_total’.

```python
ventas_por_categoria = df.groupby("categoria")["venta_total"].sum()

promedio_por_categoria = df.groupby("categoria")["venta_total"].mean()

print(ventas_por_categoria)
print(promedio_por_categoria)
```
