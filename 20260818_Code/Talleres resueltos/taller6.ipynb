```python
'''Ejercicio 2: La Calculadora de KPIs
Necesitas calcular la ganancia neta de una venta restando los costos operativos.
Crea una función llamada calcular_ganancia que reciba dos parámetros: ingreso_total y costos. 
La función debe calcular la resta de estos dos valores y retornar (return) el resultado.
Finalmente, prueba tu función con ingresos y costos insertados por consola, guardando el resultado en una variable e imprimiéndolo.'''

def calcular_ganancia(ingreso_total, costos):
    ganancia = ingreso_total - costos
    return ganancia

ingreso = float(input("Ingrese el total de ingresos: "))
costos = float(input("Ingrese el total de costos: "))

resultado = calcular_ganancia(ingreso, costos)

print("La ganancia neta es:", resultado)
```

```python
'''Ejercicio 3: El Estandarizador de Columnas (Limpieza de texto)
Los nombres de las columnas que vienen de Excel suelen venir en mayúsculas y con espacios, lo que dificulta el análisis.
Crea una función llamada limpiar_columna que reciba una cadena de texto (texto_sucio). La función debe convertir el texto
a minúsculas y reemplazar los espacios " " por guiones bajos "_" usando el método .replace().
Retorna el texto limpio. Pruébala con el texto que quieras tomándolo con el método input.'''


def limpiar_columna(texto_sucio):
    texto_limpio = texto_sucio.lower()
    texto_limpio = texto_limpio.replace(" ", "_")
    return texto_limpio


texto = input("Ingrese el nombre de la columna: ")

resultado = limpiar_columna(texto)

print(resultado)
```

```python
''' Ejercicio 4: El Auditor de Metas (Integrando condicionales)
Necesitas una herramienta que audite rápidamente si una sucursal cumplió su cuota de ventas.
Crea una función llamada auditar_meta que reciba ventas_reales y meta_esperada. Dentro de la función, 
usa un if/else. Si las ventas reales son mayores o iguales a la meta, retorna "Meta Cumplida". De lo contrario, 
retorna "Meta No Cumplida".
Pruébala con los valores numéricos tomados con el método input.'''

def auditar_meta(ventas_reales, meta_esperada):
    
    if ventas_reales >= meta_esperada:
        return "Meta Cumplida"
    
    else:
        return "Meta No Cumplida"
```

```python
ventas = float(input("Ingrese las ventas reales: "))
meta = float(input("Ingrese la meta esperada: "))

resultado = auditar_meta(ventas, meta)
print(resultado)
```

```python
'''Ejercicio 5: El Procesador de Lotes (Integrando bucles)
Tienes una lista de temperaturas extraídas de un sensor y necesitas saber cuántas de ellas superan los 30 grados para lanzar una alerta.
Instrucción: Crea una función llamada contar_alertas que reciba una lista_temperaturas. La función debe inicializar un contador en 0,
 usar un ciclo for para recorrer la lista y, si la temperatura es mayor a 30, sumar
  al contador.
Al final, debe retornar el total de alertas.
Pruébala con la lista:
[25, 32, 28, 35, 31, 22]'''

def contar_alertas(lista_temperaturas):
    contador = 0
    for temperatura in lista_temperaturas:
        if temperatura > 30:
            contador += 1
    return contador 
```

```python
temperaturas = [25, 32, 28, 35, 31, 22]
alertas = contar_alertas(temperaturas)
print("Total de alertas:", alertas)
```

```python
'''Ejercicio 6: El Filtro de Calidad
Al extraer datos de un sistema antiguo, algunas edades de clientes se registraron con números negativos o ceros por error. 
Necesitamos auditar eso.
Crea una función llamada auditar_edades que reciba una lista de números (lista_edades). La función debe iniciar un contador de 
errores en 0, usar un ciclo for para recorrer cada edad, y un condicional if para evaluar si la edad es menor o igual a 0. Si lo es,
 suma 1 al contador de errores. Al final, la función debe retornar el total de errores encontrados.
Pruébala con la lista [25, -2, 30, 0, 45, 18, -5, 0, 52].'''


def auditar_edades(lista_edades):
    contador_errores = 0

    for edad in lista_edades:
        if edad <= 0:
            contador_errores += 1

    return contador_errores
```

```python
lista_edades = [25, -2, 30, 0, 45, 18, -5, 0, 52]
errores = auditar_edades(lista_edades)
print("Total de errores encontrados:", errores)
```

```python
'''Reto Extra 2: El Generador de Correos (Función que inserta en listas)
El equipo de marketing te pasó una lista con nombres de empleados nuevos y necesitas crearles sus correos corporativos 
automáticamente y guardarlos en un nuevo registro. Crea una función llamada generar_correos que reciba una lista de nombres
(lista_nombres). Dentro de la función, crea una lista vacía llamada lista_correos. Usa un ciclo for para recorrer los nombres;
a cada nombre agrégale la terminación "@empresa.com" y conviértelo a minúsculas. Usa el método .append() para insertar este
nuevo correo en tu lista vacía. Al final, retorna la lista completa de correos. Pruébala con valores dados con input.'''

def generar_correos(lista_nombres):
    lista_correos = []

    for nombre in lista_nombres:
        correo = nombre.lower() + "@empresa.com"
        lista_correos.append(correo)

    return lista_correos

```

```python
cantidad = int(input("¿Cuántos nombres vas a ingresar? "))

lista_nombres = []

for i in range(cantidad):
    nombre = input("Ingrese un nombre: ")
    lista_nombres.append(nombre)

resultado = generar_correos(lista_nombres)

print(resultado)
```
