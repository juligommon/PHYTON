```python
#Ejercicio 10: PIN de Seguridad (Intentos controlados): 
# Define un PIN secreto (ej. 1234). 
# Pide al usuario que lo ingrese.
#  Repite el ciclo mientras el PIN sea incorrecto, pero si acierta, saluda y rompe el ciclo.
 
pin_secreto = 1234

while True:
    pin = int(input("Ingrese el PIN: "))

    if pin != pin_secreto:
        print("PIN incorrecto. Intente nuevamente.")
    else:
        print("¡PIN correcto! Bienvenido.")
        break
```

```python
'''Ejercicio 7: Sumadora de Ventas
Enunciado:
Pide valores de ventas continuos y súmalos en una variable total.
 El ciclo debe detenerse cuando el usuario ingrese un valor de 0. Imprime el total final.'''

total = 0

venta = float(input("Ingrese el valor de la venta: "))

while venta != 0:
    total = total + venta
    venta = float(input("Ingrese el valor de la venta: "))

print("Total final:", total)
```

```python
''' Ejercicio 16: Contador de Metas Alcanzadas
Dada la lista de ventas diarias: 
[800, 1200, 950, 1500]
Recorre la lista. Si la venta es mayor a 1000, imprime “Meta lograda”'''

ventas_diarias = [800, 1200, 950, 1500]

for venta in ventas_diarias:
    if venta > 1000:
        print("Meta lograda")
```

```python
'''Ejercicio 18: Limpiador de Datos
Dada la lista:
edades = [25, -1, 30, 0, 45]
Recórrela. Si la edad es menor o igual a 0, usa continue para ignorarla. Si es válida, imprime:
“Edad válida: [edad]”'''

edades = [25, -1, 30, 0, 45]

for edad in edades:
    if edad <= 0:
        continue

    print("Edad válida:", edad)
```

```python
''' Ejercicio 19: Buscador de Errores Críticos
Dada la lista de estados:
["OK", "OK", "ERROR", "OK"]
Recorre la lista; si encuentras "ERROR", imprime:
“Falla crítica detectada, deteniendo…”
y usa break para romper el ciclo.'''

estados = ["OK", "OK", "ERROR", "OK"]

for estado in estados:
    if estado == "ERROR":
        print("Falla crítica detectada, deteniendo...")
        break
```

```python
'''Ejercicio 20: El Filtro Interactivo
Crea un ciclo infinito (while True). Pide una calificación del 1 al 10.
* Si es mayor a 5, imprime “Aprobado”.
* Si es menor o igual a 5, imprime “Reprobado”.
* Si el usuario ingresa 0, rompe el ciclo con break.'''

while True:
    calificacion = int(input("Ingresa una calificación del 1 al 10 (0 para salir): "))

    if calificacion == 0:
        break

    if calificacion > 5:
        print("Aprobado")
    else:
        print("Reprobado")
```
