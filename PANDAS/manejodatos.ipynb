1. Carga y Manipulación de Datos

Contexto: Iniciamos cargando el archivo de sucursales con valores nulos y generando columnas calculadas para enriquecer el análisis básico.

```python
import pandas as pd
```

```python
df = pd.read_csv(
    'ejemplo.csv',
    sep=';',
    encoding='utf-8'
)

df
```

```python
# copia_df_or = df.copy()
# copia_df_or
```

```python
df["Cantidad"] = df["Cantidad"].fillna(1)
df["Vendedor"] = df["Vendedor"].fillna("Jeff")

df
```

```python
df["New_Col"] = "HHH"
df
```

```python
df["New_Col"] = df["Cantidad"] * df["Precio_Unitario"]
df
```

```python
# Imputamos la cantidad nula con un valor por defecto o promedio
df["Cantidad"] = df["Cantidad"].fillna(1)

# Creación de columna calculada
df["Venta_Total"] = df["Cantidad"] * df["Precio_Unitario"]

print(df[["ID_Pedido", "Producto", "Venta_Total"]].head(3))
```

2. Manipulación de Cadenas (.str)

Contexto: Estandarizamos los textos de las columnas categóricas (como convertir las ciudades a minúsculas o limpiar espacios) para asegurar que los filtros y agrupaciones no fallen por variaciones tipográficas.

```python
# Estandarizar la columna 'Ciudad' a minúsculas y quitar espacios
df["Ciudad_Limpia"] = df["Ciudad"].str.lower().str.strip()
df

# Reemplazar valores de texto en nombres de productos si fuera necesario
#df["Producto"] = df["Producto"].str.replace("Mouse", "Ratón Óptico")

#print(df[["Ciudad", "Ciudad_Limpia", "Producto"]].head(3))
```

```python
# Reemplazar valores de texto en nombres de productos si fuera necesario
df["Producto"] = df["Producto"].str.replace("Mouse", "Ratón Óptico")
df
```

```python
print(df[["Ciudad", "Ciudad_Limpia", "Producto"]].head(3))
```

3. Ordenamiento de Datos (sort_values)

Contexto: Ordenamos las transacciones comerciales de mayor a menor valor para destacar rápidamente los pedidos que generan más ingresos.

```python
df_ordenado = df.sort_values(by='New_Col', ascending=True)
df_ordenado
```

```python
df_ordenado = df.sort_values(by="New_Col", ascending=False)

df_ordenado[["Vendedor", "New_Col"]].head(1)
```

```python
# Ordenar por 'Venta_Total' de forma descendente
df_ordenado = df.sort_values(by="New_Col", ascending=False)

print("--- Transacciones ordenadas por mayor valor ---")
print(df_ordenado[["ID_Pedido", "Producto", "New_Col"]].head(3))
```

4. Agrupamiento y Agregaciones (groupby)

Contexto: Agrupamos las operaciones por dimensiones clave (por ejemplo, por Ciudad o por Vendedor) para resumir el rendimiento total del negocio.

En la parte inferior de la imagen empieza el siguiente código:

```python
resumen_ciudad = df.groupby('Ciudad_Limpia')['New_Col'].sum()
resumen_ciudad
```

```python
# Agrupar por Ciudad sumando las ventas totales y promediando los precios
resumen_ciudad = df.groupby("Ciudad").agg(
    Ingreso_Total=("Venta_Total", "sum"),
    Precio_Promedio=("Precio_Unitario", "mean")
).reset_index()
```

```python
resumen_ciudad = (df.groupby(['Ciudad_Limpia','Producto'])['New_Col']
                  .sum()
                  .sort_values(ascending=False)
                 )

resumen_ciudad
#resumen_ciudad
```
